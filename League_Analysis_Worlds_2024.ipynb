```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt 
import seaborn as sns

```

```python
%%sql

SELECT 
    ad.teamname,
    ad.playername, 
    ad.champion,
    COUNT(ad.champion) AS champion_count,
    ROUND(AVG(ad.cspm),2) AS CSPM,
    SUM(ad.firstblood) AS FB,
    ROUND(SUM(ad.firstblood) / COUNT(ad.game)*100,1) AS FB_perce,
    ROUND(avg(ad.dpm),1) AS avg_champ_damage,
    ROUND(avg(ad.damageshare)*100,1) AS avg_champ_damageshare,
    ROUND(SUM(CASE WHEN ad.result = 1 THEN 1 ELSE 0 END) / COUNT(ad.result) *100, 2) AS win_rate_per,
    AVG(ad.gamelength) AS avg_game_length,
    STDDEV(ad.gamelength) AS std_game_length,
    ROUND(IF(SUM(ad.deaths + ad.assists) = 0, NULL, SUM(ad.kills + ad.assists) / SUM(ad.deaths)), 2) AS KDA,
    ROUND(SUM(IFNULL(ad.towers, 0)) / COUNT(ad.game), 1) AS total_towers_per_game,
    ROUND(SUM(IFNULL(ad.dragons, 0)) / COUNT(ad.game), 1) AS total_dragons_per_game
FROM 
    Worlds_2024.all_data_2024 ad
WHERE 
    ad.patch = '14.18' 
    AND ad.league LIKE '%WLD%'
    AND (ad.teamname LIKE '%Bili%' OR ad.teamname LIKE '%T1%')
GROUP BY 
    ad.playername, 
    ad.champion,
    ad.firsttower,
    ad.teamname,
    ad.towers
ORDER BY 
    ad.playername DESC;
```
